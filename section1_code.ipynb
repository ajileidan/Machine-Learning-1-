```python
from PIL import Image
import glob
import numpy as np
import os
import re


def main():

    src = "C:\\Users\\ammar\\Desktop\\A1_Ammar_Jileidan_40450780\\pgms/*.pgm"

    # open every PGM file 
    for file in glob.glob(src):

        filename = os.path.basename(file)
        # to extract the label
        result =  re.findall(r'[A-Za-z]+|\d+', filename)
        label = result[0]
        # to extract the index
        index = result[1].split(".")[0]
        
        img = Image.open(file)

        arr = np.array(img)
        # convert from grayscale into 0s and 1s
        arr = np.where(arr < 128 , 0, 1)
        # formating CSV file name
        outfilename = "40450780" + "_" + label + "_" + index.zfill(2) + ".csv"
        
        np.savetxt(os.path.join("C:\\Users\\ammar\\Desktop\\AI a1\\csvs", outfilename), arr, delimiter=',', fmt='%d' )

        

main()
        
```
